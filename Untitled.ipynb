```julia
# Overfitting: simulación y gráficos
using Random, Statistics, LinearAlgebra, Plots

# -----------------------------
# 1) Parámetros y DGP
# -----------------------------
Random.seed!(123)
n = 1_000
σ = 1.0

# DGP: y = 1*x + ε, sin intercepto
x = randn(n)
ε = σ .* randn(n)
y = x .+ ε

# lista de cantidades de features (polinomios de x)
p_list = [1, 2, 5, 10, 20, 50, 100, 200, 500, 1000]

# contenedores
R2_full   = Float64[]
R2_adj    = Float64[]
R2_oos    = Float64[]

# utilidades
make_design(xvec, p) = hcat([xvec .^ k for k in 1:p]...)  # sin intercepto

function r2_and_adj(y, ŷ, p)
    sst = sum((y .- mean(y)).^2)
    sse = sum((y .- ŷ).^2)
    r2  = 1 - sse/sst
    r2adj = 1 - (1 - r2) * (length(y) - 1) / (length(y) - p)
    return r2, r2adj
end

function r2_oos(ytest, yhat_test)
    sst_test = sum((ytest .- mean(ytest)).^2) # baseline: media del set de prueba
    sse_test = sum((ytest .- yhat_test).^2)
    return 1 - sse_test/sst_test
end

```

```julia
# -----------------------------
# 2) Bucle por número de features
# -----------------------------
for p in p_list
    # diseño
    X = make_design(x, p)

    # --- Full sample ---
    β̂ = X \ y
    ŷ = X * β̂
    r2, r2adj = r2_and_adj(y, ŷ, p)
    push!(R2_full, r2)
    push!(R2_adj,  r2adj)

    # --- Train/Test 75/25 ---
    idx = shuffle(1:n)
    ntrain = round(Int, 0.75n)
    train, test = idx[1:ntrain], idx[ntrain+1:end]

    Xtr, ytr = X[train, :], y[train]
    Xte, yte = X[test,  :], y[test]

    β̂_tr = Xtr \ ytr
    ŷ_te = Xte * β̂_tr
    push!(R2_oos, r2_oos(yte, ŷ_te))
end
```

```julia
# -----------------------------
# 3) Gráficos (tres separados)
# -----------------------------
default(marker=:circle)

p1 = plot(p_list, R2_full, xlabel="Número de features", ylabel="R²",
          title="R² en muestra completa", legend=false, xscale=:log10)
p2 = plot(p_list, R2_adj, xlabel="Número de features", ylabel="R² ajustado",
          title="R² ajustado en muestra completa", legend=false, xscale=:log10)
p3 = plot(p_list, R2_oos, xlabel="Número de features", ylabel="R² fuera de muestra",
          title="R² out-of-sample (75/25)", legend=false, xscale=:log10)

display(p1); display(p2); display(p3)

# Opcional: guardar
# savefig(p1, "R2_full.png")
# savefig(p2, "R2_adj.png")
# savefig(p3, "R2_oos.png")
```
